# Held-out link-prediction results

Select the task directory, datasets and expected seeds below. Configurations are
selected by mean validation MRR over the same complete seed set. Every displayed
MRR/Hits value is a held-out test metric from the validation-selected checkpoint.
Missing test metrics, missing seeds and duplicate runs raise an error instead of
falling back to validation scores. Static/temporal summary logs assume the
trainers were run with early stopping enabled.


```python
from pathlib import Path
import sys

REPO_ROOT = Path.cwd().resolve()
if REPO_ROOT.name == "notebooks":
    REPO_ROOT = REPO_ROOT.parent
sys.path.insert(0, str(REPO_ROOT / "notebooks"))
from link_result_utils import collect_runs, summarize_runs, latex_table

TASK_DIR = REPO_ROOT / "brightkite"  # or REPO_ROOT / "amazon"
DATASETS = {
    "brightkite": ["brightkite", "gowalla"],
    "amazon": ["clothing", "electronics", "movies"],
}[TASK_DIR.name]
EXPECTED_SEEDS = (0, 1, 2)
```

```python
runs = collect_runs(TASK_DIR, DATASETS)
summary = summarize_runs(runs, EXPECTED_SEEDS)
for row in summary:
    print(row["dataset"], row["method"], "validation MRR:", row["validation_mrr_mean"])
    print("  selected configuration:", row["configuration"], "seeds:", row["seeds"])
for dataset in DATASETS:
    print(latex_table(summary, dataset))
```

```python
import json

OUTPUT_DIR = REPO_ROOT / "tables"
OUTPUT_DIR.mkdir(exist_ok=True)
summary_path = OUTPUT_DIR / f"link_{TASK_DIR.name}.json"
summary_path.write_text(json.dumps(summary, indent=2) + "\n")
for dataset in DATASETS:
    (OUTPUT_DIR / f"link_{dataset}.tex").write_text(latex_table(summary, dataset) + "\n")
print("Wrote", summary_path)
```
